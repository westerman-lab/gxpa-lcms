# Pathway enrichment of GxM interaction statistics (mummichog)

Takes the metabolome-wide interaction statistics from `02b_analysis` and asks which
metabolic pathways are enriched among the features that interact with each SNP. Because
most features are unannotated, pathway-level inference works directly from *m/z* and
retention time and does not depend on compound identification.

**Inputs** (both written to `results/` by `02b_analysis`):

* `results/GxM_results.csv` -- one row per lane per metabolite (a lane is a declared (SNP, outcome, exposure) in `anchors.csv`)
* `results/met_info-mesa.csv` -- *m/z*, retention time and LC-MS method per feature

**Output**: `results/mummichog_pathways.csv`, one row per pathway per lane per mode, keyed by `lane_id`.

## Prerequisites

Mummichog is fussy. It needs **version 2.6.1 exactly** (later versions have a bug) and
**Python 3.8**:

```
pyenv install 3.8 && pyenv global 3.8 && pip install mummichog==2.6.1
```

Terra images do not carry Python 3.8, so this notebook is usually easier to run locally.
The preflight cell below prints `mummichog -h`; option letters shifted between 2.x
releases, so check them against the parameters in the setup cell before a long run.

````r
library(data.table)

# ---- Parameters ------------------------------------------------------------
PPM     <- 10L                 # Instrument mass accuracy, ppm     (mummichog -u)
PERMS   <- 1000L               # Permutations for the null         (mummichog -p)
                               # 100 left p-values near 0.05 unstable: a borderline
                               # pathway flipped significance between two runs of the
                               # same input. Strong p-values are analytic and were
                               # already stable; this settles the borderline ones.
NETWORK <- 'human_mfn'         # Metabolic network                 (mummichog -n)

# Ionization mode per LC-MS method. This matters: negative-mode features pushed
# through positive-mode adduct rules get mis-assigned masses, so each mode is
# built and run as a separate job rather than pooling all four methods.
METHOD_MODE <- c('C8-pos'    = 'generic_positive',
                 'HILIC-pos' = 'generic_positive',
                 'C18-neg'   = 'negative',
                 'Amide-neg' = 'negative')

in_dir  <- 'scratch/mummichog_input'
out_dir <- 'scratch/mummichog_output'
for (d in c(in_dir, out_dir, 'results')) dir.create(d, recursive=TRUE, showWarnings=FALSE)

ws_bucket <- Sys.getenv('WORKSPACE_BUCKET',
                        unset='gs://fc-secure-4a392455-5587-4d6f-b8bd-01a1f834ae63')

# --- Lanes ---------------------------------------------------------------------
# A lane is one declared (G, Y, E) triple: a SNP, the outcome, and the exposure whose
# interaction with it is followed up. anchors.csv declares them, and every consumer
# (01c, 02a, 02b, 02c, 03c, manuscript.Rmd) reads that one file, so adding or parking a
# lane is one row there rather than an edit in each notebook. On Terra only *.ipynb
# reaches the VM, so the copy read is the one published to the workspace bucket.

# The trait a variable measures, ignoring how it was derived: bmi, bmi_covariate and
# bmi_wins are all BMI. Extends the suffix convention 01c's observed() uses.
trait <- \(v) sub('_(log|wins|trunc|covariate)$', '', v)

read_lanes <- function(bucket) {
  lanes <- fread(cmd = paste0('gcloud storage cat ', bucket, '/anchors.csv'))
  miss  <- setdiff(c('lane_id','G','Y','E','rsid','gene','label','include'), names(lanes))
  if (length(miss))
    stop('anchors.csv is missing column(s) ', paste(miss, collapse=', '),
         ' -- has the current version been published to the bucket? See the README.',
         call.=FALSE)
  if (!'extra_covars' %in% names(lanes)) lanes[, extra_covars := '']
  lanes[, extra_covars := fifelse(is.na(extra_covars), '', as.character(extra_covars))]
  lanes[, include := as.logical(include)]
  # A control lane is screened like any other but never reported: it exists to give the
  # screen's cross-cohort statistics a null distribution. 02c, both reports and every
  # per-lane QC table filter on role. Defaulted, so an anchors.csv predating the column
  # still reads as all-primary.
  if (!'role' %in% names(lanes)) lanes[, role := 'primary']
  lanes[, role := fifelse(is.na(role) | !nzchar(role), 'primary', as.character(role))]
  if (!all(lanes$role %in% c('primary', 'control')))
    stop('anchors.csv: role must be primary or control; got ',
         paste(setdiff(unique(lanes$role), c('primary','control')), collapse=', '), call.=FALSE)
  if (anyDuplicated(lanes$lane_id))
    stop('anchors.csv: lane_id must be unique; duplicated: ',
         paste(unique(lanes$lane_id[duplicated(lanes$lane_id)]), collapse=', '), call.=FALSE)
  # Unique by trait, not just by name, so ensure_lane_id() below can match on traits
  # without ambiguity.
  if (anyDuplicated(lanes[, .(G, trait(Y), trait(E))]))
    stop('anchors.csv: two lanes declare the same SNP, outcome trait and exposure trait.', call.=FALSE)
  # A lane with one trait on both sides would fit e.g. bmi ~ G*bmi_covariate. Compared by
  # trait rather than by name, so that case is caught and not only bmi ~ G*bmi.
  same <- lanes[trait(Y) == trait(E), lane_id]
  if (length(same))
    stop('anchors.csv: outcome and exposure are the same trait in ',
         paste(same, collapse=', '), '. A lane cannot use one trait on both sides.', call.=FALSE)
  lanes[]
}

# Covariates for one model: the shared set plus a lane's own additions, minus anything
# measuring that model's outcome or exposure. By trait, because a name-level setdiff
# misses exactly the case that matters -- bmi_covariate as a precision covariate in a
# lane whose outcome is bmi survives setdiff(covars, c('bmi','mvpa_wins')).
lane_covars <- function(covars, Y, E = NULL, extra = '') {
  extra <- if (length(extra) == 1 && !is.na(extra) && nzchar(extra))
             strsplit(trimws(extra), '[ ;,]+')[[1]] else character()
  cs <- unique(c(covars, extra))
  cs[!trait(cs) %in% trait(c(Y, E))]
}

# Results checkpointed before lanes existed carry (G, Y, E) but no lane_id. Attach it
# rather than refusing them, so a notebook still reads older results without re-fitting.
# Matched on the SNP and the outcome and exposure TRAITS, so a result written when the
# CETP exposure was still named bmi attaches to the lane that now names it bmi_covariate.
# Rows matching no declared lane keep lane_id = NA.
ensure_lane_id <- function(dt, lanes) {
  if ('lane_id' %in% names(dt) || !all(c('G','Y','E') %in% names(dt))) return(dt)
  key <- lanes[, .(lane_id, G, Y_trait = trait(Y), E_trait = trait(E))]
  key[copy(dt)[, `:=`(Y_trait = trait(Y), E_trait = trait(E))], on=.(G, Y_trait, E_trait)
    ][, c('Y_trait', 'E_trait') := NULL][]
}
````

## Preflight

Check the binary and the inputs up front, so a missing dependency surfaces here rather
than part-way through the run loop.

````r
# Resolve the mummichog binary by absolute path. PATH is unreliable here: the R kernel
# inherits the PATH Jupyter was started with, so `export PATH=...` in a terminal is never
# visible to this notebook -- not even after a kernel restart, which re-forks from the same
# Jupyter process. Set MUMMICHOG_BIN to override; otherwise the usual isolated-install
# locations are checked before falling back to PATH.
resolve_mummichog <- function() {
  cand <- c(Sys.getenv('MUMMICHOG_BIN'),
            path.expand('~/envs/mummichog/bin/mummichog'),
            path.expand('~/.local/bin/mummichog'),
            path.expand('~/.local/share/uv/tools/mummichog/bin/mummichog'),
            unname(Sys.which('mummichog')))
  cand <- cand[cand != '' & file.exists(cand)]
  if (length(cand)) cand[1] else ''
}
mummichog_bin <- resolve_mummichog()

if (mummichog_bin == '') {
  message('mummichog NOT on PATH -- inputs below will still be built, but the run cell will skip.\n',
          'It needs Python 3.8 exactly: mummichog 2.6.1 pins networkx 1.11, which imports\n',
          'fractions.gcd -- removed in Python 3.9. Install it ISOLATED so its ancient deps\n',
          'do not shadow the notebook environment (a --user install breaks scikit-image):\n',
          '  conda create -y -p ~/envs/mummichog python=3.8\n',
          '  PIP_USER=false ~/envs/mummichog/bin/pip install mummichog==2.6.1\n',
          'PIP_USER=false is required on Terra, which sets PIP_USER=true globally: without\n',
          'it pip installs into ~/.local even when you call the env\'s own pip, which both\n',
          'pollutes the shared environment and puts the launcher somewhere unexpected.\n',
          'No PATH or symlink step is needed -- this notebook resolves the binary by\n',
          'absolute path. Verify with:  ~/envs/mummichog/bin/mummichog   (no args = usage).')
} else {
  cat('mummichog:', mummichog_bin, '\n\n')
  system(paste(shQuote(mummichog_bin), '-h'))   # Confirm the option letters used in the run cell.
}

# Pull the inputs rather than assuming an earlier session left them on disk.
grab_file_if_not_extant <- \(f) if(!file.exists(f)) {
  dir.create(dirname(f), recursive=TRUE, showWarnings=FALSE)
  system(paste0('gcloud storage cp ', ws_bucket, '/', f, ' ', f))
}
grab_file_if_not_extant('results/GxM_results.csv')                                # 02b
grab_file_if_not_extant('data/derived/metabolomics/met_info-mesa.csv')            # 01a

# 02b writes both inputs into results/; fall back to the raw metadata location.
gxm_path <- 'results/GxM_results.csv'
met_path <- if (file.exists('results/met_info-mesa.csv')) 'results/met_info-mesa.csv' else
                                                          'data/derived/metabolomics/met_info-mesa.csv'
for (f in c(gxm_path, met_path))
  if (!file.exists(f)) stop('Missing input: ', f, ' -- run 02b_analysis first.')
lanes <- read_lanes(ws_bucket)

cat('\ninteraction stats :', gxm_path,
    '\nfeature metadata  :', met_path, '\n')
````

## Build one input per lane x ionization mode

A *lane* is one (SNP, outcome, exposure) combination declared in `anchors.csv` and screened in
`02b`. Each lane is split by ionization mode, so the number of jobs is lanes x 2. Mummichog's
input is tab-delimited: *m/z*, retention time, p-value, t-statistic, feature id.

````r
gxm <- ensure_lane_id(fread(gxm_path), lanes) # A checkpoint written before lanes carries no lane_id.
if (gxm[, any(is.na(lane_id))])
  message('Dropping ', gxm[is.na(lane_id), .N], ' row(s) that match no lane declared in anchors.csv.')
gxm <- gxm[!is.na(lane_id)]
# Control lanes are screened by 02b to give the cross-cohort statistics a null distribution;
# they are not results, and a pathway run over them would be 2 jobs each for nothing.
n_ctrl <- gxm[lane_id %in% lanes[role == 'control', lane_id], uniqueN(lane_id)]
if (n_ctrl) message('Skipping ', n_ctrl, ' control lane(s): pathway analysis is for reported lanes only.')
gxm <- gxm[lane_id %in% lanes[role != 'control', lane_id]]
met <- fread(met_path)[, .(unique_met_id, Method, MZ, RT)]

mc_input <- met[gxm, on=.(unique_met_id = M)   # Attach m/z and RT to each tested feature.
  ][!is.na(MZ) & !is.na(RT) & !is.na(p)        # Mummichog needs all three.
  ][, mode := METHOD_MODE[Method]
  ][!is.na(mode)                               # Drop methods with no mode mapping.
  ][, tag  := paste(lane_id, mode, sep='-')
]

# Report what is excluded and why, rather than dropping features silently.
excluded <- met[gxm, on=.(unique_met_id = M)][is.na(MZ) | is.na(RT) | is.na(p)]
if (nrow(excluded))
  print(excluded[, .(excluded_features = .N), by=.(Method)][order(-excluded_features)])
cat('\nNote: Amide-neg is a targeted MRM method and carries no retention time',
    '(01a moves it to MRM_Transition), so those features cannot be used here.\n\n')

cat('usable features:', nrow(mc_input), 'of', nrow(gxm), 'tested\n\n')
print(mc_input[, .(features = .N), by=.(Method, mode)][order(mode, Method)])

inputs <- unique(mc_input[, .(tag, lane_id, Y, E, G, mode)])
inputs[, file := file.path(in_dir, paste0(tag, '.tsv'))]
for (i in seq_len(nrow(inputs))) {
  fwrite(mc_input[tag == inputs$tag[i], .(MZ, RT, p, t, unique_met_id)],
         inputs$file[i], sep='\t')
}
inputs[, features := mc_input[, .N, by=tag][inputs, on='tag', N]]
inputs[, min_p    := mc_input[, min(p), by=tag][inputs, on='tag', V1]]

inputs[]
````

## Run

Each job runs inside its own directory, so mummichog's timestamped output folder lands
somewhere predictable -- no guessing at generated names afterwards. Console output is
captured per job; if a run fails, read `scratch/mummichog_output/<tag>/mummichog.stdout.txt`
(a wrong option letter shows up here first).

````r
run_mummichog <- function(input_file, tag, mode) {
  wd <- file.path(out_dir, tag)
  unlink(wd, recursive=TRUE); dir.create(wd, recursive=TRUE, showWarnings=FALSE)
  cmd <- paste('cd', shQuote(wd), '&&', shQuote(mummichog_bin),
               '-f', shQuote(normalizePath(input_file)),
               '-o', 'out',
               '-m', mode,
               '-u', PPM,
               '-p', PERMS,
               '-n', NETWORK,
               '> mummichog.stdout.txt 2>&1')
  status  <- system(cmd)
  pw_file <- list.files(wd, pattern='pathwayanalysis.*\\.tsv$', recursive=TRUE)
  data.table(tag = tag, exit_status = status, produced_pathways = length(pw_file) > 0)
}

if (mummichog_bin != '') {
  t0 <- Sys.time()
  run_log <- inputs[, .(tag, lane_id, mode)][
    rbindlist(Map(run_mummichog, inputs$file, inputs$tag, inputs$mode)), on='tag']
  print(Sys.time() - t0)
  fwrite(run_log, 'results/mummichog_run_log.csv')
  if (any(!run_log$produced_pathways))
    message('Some jobs produced no pathway table -- check their mummichog.stdout.txt.')
  run_log[]
} else {
  message('mummichog not found -- skipping. Inputs are built and ready in ', in_dir)
}
````

## Collect

The pathway table is located by pattern rather than exact filename, since mummichog's
output naming varies across versions.

````r
collect_pathways <- function(tag) {
  f <- list.files(file.path(out_dir, tag), pattern='pathwayanalysis.*\\.tsv$',
                  recursive=TRUE, full.names=TRUE)
  if (length(f) == 0) return(NULL)
  fread(f[1])[, tag := tag]
}

pathways <- rbindlist(lapply(inputs$tag, collect_pathways), fill=TRUE)

if (nrow(pathways)) {
  setnames(pathways, names(pathways),          # Tidy mummichog's spaces/parentheses.
           gsub('_+$', '', gsub('[^A-Za-z0-9]+', '_', names(pathways))))
  pathways <- inputs[, .(tag, lane_id, Y, E, G, mode)][pathways, on='tag']
  fwrite(pathways, 'results/mummichog_pathways.csv')
  cat('pathway rows:', nrow(pathways), 'across', uniqueN(pathways$tag), 'jobs\n')
} else {
  message('No pathway tables found -- nothing collected.')
}

head(pathways, 20)
````

## Top pathways

````r
if (nrow(pathways)) {
  # Mummichog's p-value column has been named several things across versions.
  pcol <- grep('^p_value|^p_val|^pvalue|^p$', names(pathways), value=TRUE, ignore.case=TRUE)[1]
  if (is.na(pcol)) {
    message('Could not identify a p-value column; showing raw table. Columns: ',
            paste(names(pathways), collapse=', '))
    head(pathways, 20)
  } else {
    pathways[order(get(pcol))
      ][, head(.SD, 5), by=.(lane_id, mode)      # Top 5 pathways per job.
    ]
  }
}
````
